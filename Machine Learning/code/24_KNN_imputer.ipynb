```python
import pandas as pd 
import numpy as np 

from sklearn.model_selection import train_test_split
from sklearn.impute import KNNImputer, SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df= pd.read_csv('/Users/prajwalsubedi/Desktop/Data science/Data Sets/train1.csv')[['Age','Pclass','Fare','Survived']]
```

```python
df.head()
```

```python
df.isnull().mean()
```

```python
df.isnull().sum()
```

```python
x =df.drop(columns=['Survived'])
y= df['Survived']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.head()
```

```python
knn= KNNImputer(n_neighbors=5,weights='distance')
x_train_trf= knn.fit_transform(x_train)
x_test_trf= knn.fit_transform(x_test)
```

```python
lr = LogisticRegression()
lr.fit(x_train_trf,y_train)
y_pred = lr.predict(x_test_trf)
accuracy_score(y_test,y_pred)
```

```python
si = SimpleImputer()
x_train_trf1= si.fit_transform(x_train)
x_test_trf1= si.fit_transform(x_test)
```

```python
lr = LogisticRegression()
lr.fit(x_train_trf1,y_train)
y_pred2 = lr.predict(x_test_trf1)
accuracy_score(y_test,y_pred2)
```
