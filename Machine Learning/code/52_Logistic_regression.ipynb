```python
def perceptron(x,y):
    
    m = []
    b = []
    
    x = np.insert(x,0,1,axis=1)
    weights = np.ones(x.shape[1])
    lr = 0.1
    
    for i in range(200):
        j = np.random.randint(0,100)
        y_hat = step(np.dot(x[j],weights))
        weights = weights + lr*(y[j]-y_hat)*x[j]
        
        m.append(-(weights[1]/weights[2]))
        b.append(-(weights[0]/weights[2]))
        
    return m,b
```

```python
m,b = perceptron(x,y)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

fig, ax = plt.subplots(figsize=(9, 5))

x_i = np.arange(-3, 3, 0.1)

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

# Create figure
fig, ax = plt.subplots(figsize=(9, 5))

x_i = np.arange(-3, 3, 0.1)

# Scatter plot
ax.scatter(x[:, 0], x[:, 1], c=y, cmap='winter', s=100)

# Initial regression line
line, = ax.plot(x_i, m[0] * x_i + b[0], 'r-', linewidth=2)

ax.set_xlim(-3, 3)
ax.set_ylim(-3, 3)

def update(i):
    line.set_ydata(m[i] * x_i + b[i])
    ax.set_xlabel(f'Epoch: {i + 1}')
    return (line,)

# Keep the animation object alive
anim = FuncAnimation(
    fig,
    update,
    frames=len(m),
    interval=100,
    repeat=True,
    blit=False,
    cache_frame_data=False
)

# Render as HTML; do NOT call plt.close(fig)
html_animation = anim.to_jshtml()
display(HTML(html_animation))
```
