```python
from sklearn.datasets import make_regression
import numpy as np 
```

```python
x,y= make_regression(n_samples=4,n_features=1, n_informative=1, n_targets=1, noise=80, random_state=13)
```

```python
import matplotlib.pyplot as plt 
plt.scatter(x,y)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
reg= LinearRegression()
reg.fit(x,y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='red')
```

slope is constant with value 78.35
assume starting value of b be 0

```python
y_pred =((78.35*x)+0).reshape(4)
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='red',label='ols')
plt.plot(x,y_pred,color='purple',label='b=0')
plt.legend()
plt.show(
)
```

```python
m=78.35
b=0

loss_slope=-2* np.sum(y-m*x.ravel()-b)
loss_slope
```

```python
# let learning rate=0.01
lr=0.1
step_size=loss_slope*lr
step_size
```

```python
#  finding new b
b= b-step_size
b
```

```python
y_pred1=((78.35*x)+b).reshape(4)
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='red',label='ols')
plt.plot(x,y_pred1,color='purple',label='b={}'.format(b))
plt.plot(x,y_pred,color='pink',label='b=0')
plt.legend()
plt.show()
```

```python
b=-100
m=78.35
lr=0.1

epochs=10
for i in range(epochs):
    loss_slope=-2*np.sum(y-m*x.ravel()-b)
    b=b-(lr*loss_slope)
    y_pred=m*x+b
    plt.plot(x,y_pred)

plt.scatter(x,y)
```
