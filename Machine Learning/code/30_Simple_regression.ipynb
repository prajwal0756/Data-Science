```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

Linear regression is process of drawing a best fit line in a sort of linear data which passes closely through all datas.

```python
df = pd.read_csv('/Users/prajwalsubedi/Desktop/Data science/Data Sets/placement_simple.csv')
```

```python
df.sample()
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.xlabel('cgpa')
plt.ylabel('package')
```

```python
x = df.iloc[:,0:1]
y= df.iloc[:,-1]
```

```python
x
```

```python
y
```

```python
 x_train,x_test,y_train,y_test= train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
lr = LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
y_test
```

```python
lr.predict(x_test.iloc[2].values.reshape(1,1))
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.plot(x_train,lr.predict(x_train),color='red')
plt.xlabel('cgpa')
plt.ylabel('package')
```
